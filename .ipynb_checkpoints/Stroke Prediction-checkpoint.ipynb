```python
work_type = data['work_type'].value_counts()

plt.figure(figsize = (8,6))
plt.bar(work_type.index , work_type.values) # x and y
plt.xlabel("work_type")
plt.ylabel("count")
plt.show()
```

```python
Residence_type = data['Residence_type'].value_counts()

plt.figure(figsize = (8,6))
plt.bar(Residence_type .index , Residence_type .values) # x and y
plt.xlabel("Residence_type")
plt.ylabel("count")
plt.show()
```

```python
ever_married = data['ever_married'].value_counts()

plt.figure(figsize = (8,6))
plt.bar(ever_married.index , ever_married.values) # x and y
plt.xlabel("ever_married")
plt.ylabel("count")
plt.show()
```

```python
smoking_status = data['smoking_status'].value_counts()

plt.figure(figsize = (8,6))
plt.bar(smoking_status.index , smoking_status.values) # x and y
plt.xlabel("smoking_status")
plt.ylabel("count")
plt.show()
```

```python
numeric_cols = data.select_dtypes(include = ["float64" , "int64"]).columns

for col in numeric_cols :
    plt.figure(figsize=(8,6))
    sns.displot(data = data[col], kde = True , color = "orange")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.title(f"Histogram of {col}")
    plt.show()
```
